# House Price Prediction - Data Preprocessing
**Person B: Feature Engineering & Preprocessing**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from house_price_predictor.data.loader import DataLoader
from house_price_predictor.data.preprocessor import DataPreprocessor
print('✓ Imports successful')
```

```python
# Load train and test data
loader = DataLoader(data_dir='data/raw')
train, test = loader.load_data()
loader.get_data_info()
```

```python
# Combine train and test for unified preprocessing
combined = loader.combine_datasets()
print(f'Combined shape: {combined.shape}')
print(f'Missing values: {combined.isnull().sum().sum()} cells')
```

```python
# Initialize preprocessor
preprocessor = DataPreprocessor(combined)

# Handle missing values - fill NAs with mean/median or 'None'
preprocessor.handle_missing_values()
print(f'✓ After filling NAs: {preprocessor.data.isnull().sum().sum()} cells remaining')
```

```python
# Remove outliers (mainly from GrLivArea)
initial_rows = len(preprocessor.data)
preprocessor.remove_outliers()
removed = initial_rows - len(preprocessor.data)
print(f'✓ Outliers removed: {removed} rows')
```

```python
# Encode categorical variables to numeric
preprocessor.encode_categorical()
print(f'✓ All features now numeric')
```

```python
# Transform heavily skewed features using log(x+1)
preprocessor.transform_skewed_features(threshold=0.75)
print(f'✓ Skewed features log-transformed')
```

```python
# Create train/validation/test split
train_split, val_split, test_split = preprocessor.create_train_validation_split()
print(f'Train: {train_split.shape}')
print(f'Validation: {val_split.shape}')
print(f'Test: {test_split.shape}')
```

```python
# Save processed datasets to disk
train_split.to_csv('data/processed/train.csv', index=False)
val_split.to_csv('data/processed/val.csv', index=False)
test_split.to_csv('data/processed/test.csv', index=False)
print('✓ Processed data saved to data/processed/')
```

```python
# Visualize SalePrice distribution after preprocessing
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Before (log-transformed)
axes[0].hist(train_split['SalePrice'], bins=50, edgecolor='black')
axes[0].set_xlabel('SalePrice (log-transformed)')
axes[0].set_ylabel('Frequency')
axes[0].set_title('SalePrice Distribution')

# Check feature count
axes[1].bar(['Features'], [train_split.shape[1]])
axes[1].set_ylabel('Count')
axes[1].set_title(f'Total Features: {train_split.shape[1]}')

plt.tight_layout()
plt.savefig('results/plots/preprocessing_summary.png', dpi=100)
plt.show()
print('✓ Plot saved to results/plots/preprocessing_summary.png')
```
